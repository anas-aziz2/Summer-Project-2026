```python
import numpy as np
from LJ13.mcmc import read_xyz, run_chain

SEED = 2                    # change per chain
OUT = "LJ13_chain2"         # output prefix
N_STEPS = 20000000
TEMPERATURE = 0.10
MAX_DISPLACEMENT = 0.043
BURN_IN = 20000
STRIDE = 4000

coords, _ = read_xyz("LJ13.xyz")

configs, energy, acc = run_chain(
    coords,
    n_steps=N_STEPS,
    seed=SEED,
    temperature=TEMPERATURE,
    max_displacement=MAX_DISPLACEMENT,
    burn_in=BURN_IN,
    stride=STRIDE,
)

np.save(f"{OUT}_configs.npy", configs)
np.save(f"{OUT}_energy.npy", energy)
print(f"Acceptance: {acc*100:.2f}%  |  saved {len(configs)} configs")
```
