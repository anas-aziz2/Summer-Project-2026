```python
import torch
import torch.nn.functional as F
import numpy as np
```

```python
n_pairs = 10000


for d in range(2, 50):

    a = torch.randn(n_pairs, d)
    b = torch.randn(n_pairs, d)

    cos_values = F.cosine_similarity(a, b, dim=1)

    mean_cos = cos_values.mean().item()
    std_cos = cos_values.std().item()

    print(
        f"d = {d:2d}, "
        f"mean = {mean_cos:.4f}, "
        f"std = {std_cos:.4f}"
    )
```

You were right!
The test shows that higher dimensional vectors are more likely to be orthogonal to each other.
The standard deviation (spread) decreases as we increase the dimensions
